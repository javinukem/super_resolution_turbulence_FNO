```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import sys
import os

sys.path.append(os.path.abspath("../.."))  # This brings 'src' into the path
```

```python
df = pd.read_csv("../../experiments/cfno_2/experiment_2/fno_eval_results.csv")
x = "modes"
plt.figure(figsize=(8, 6))
sns.lineplot(data=df, x=x, y="val_loss", hue="n_operator_blocks", marker="o")

# Ensure x-axis only shows integer ticks
x_ticks = sorted(df[x].unique())
plt.xticks(x_ticks)
plt.yscale("log")
plt.xlabel(x)
plt.ylabel("Validation Loss")
plt.title("Validation Loss vs Residual Blocks (Grouped by Modes)")
plt.grid(False)
plt.tight_layout()
plt.show()
```
