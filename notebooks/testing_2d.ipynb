```python
import yaml
config = yaml.safe_load(open('config.yaml', 'r'))

import os
os.environ["CUDA_VISIBLE_DEVICES"] = config['gpu']
import torch
from src.model.models import CNN
from src.dataloader.dataloader_3d import dataset_sr
from src.dataloader.dataloader_2d import dataset_sr_2d

from torch.utils.data import random_split
import numpy as np
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
dataset = dataset_sr_2d(max_samples=30)
train_size = int(0.8 * len(dataset))
test_size = len(dataset) - train_size
train_dataset, test_dataset = random_split(dataset, [train_size, test_size])
```

```python
from matplotlib.colors import LogNorm
import matplotlib.pyplot as plt
import random
```

```python

hr_state_test, lr_state_test_tensor = test_dataset[random.randint(0, len(test_dataset))]
hr_state_test, lr_state_test = hr_state_test.numpy(), lr_state_test_tensor.numpy()

fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# plot cuts at this z level
z_level = 2
for i, state in enumerate([lr_state_test, hr_state_test]):
    axes[i,0].imshow(state[0, :, :].T, origin = "lower", extent = [0, 1, 0, 1], norm = LogNorm())
    axes[i,0].set_title("Density")

    axes[i,1].imshow(np.sqrt(state[1, :, :]**2 + state[2, :, :]**2).T, origin = "lower", extent = [0, 1, 0, 1])
    axes[i,1].set_title("Velocity")

    axes[i,2].imshow(state[4, :, :].T, origin = "lower", extent = [0, 1, 0, 1], norm = LogNorm())
    axes[i,2].set_title("Pressure")

    # equal aspect ratio
    axes[i,0].set_aspect('equal', 'box')
    axes[i,1].set_aspect('equal', 'box')
    axes[i,2].set_aspect('equal', 'box')
```
