```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import sys
import os

sys.path.append(os.path.abspath(".."))  # This brings 'src' into the path
```

```python
df = pd.read_csv("eval_results.csv")

df_filtered = df[df["kernel_size"] == 3]
x = "res_blocks"
plt.figure(figsize=(8, 6))
sns.lineplot(data=df_filtered, x=x, y="val_loss", hue="features", marker="o")

# Ensure x-axis only shows integer ticks
x_ticks = sorted(df[x].unique())
plt.xticks(x_ticks)
plt.yscale("log")
plt.xlabel(x)
plt.ylabel("Validation Loss")
plt.title("Validation Loss vs Residual Blocks (Grouped by Modes)")
plt.grid(False)
plt.tight_layout()
plt.show()
```

```python
x = "res_blocks"
plt.figure(figsize=(8, 6))
sns.lineplot(data=df, x=x, y="val_loss", hue="kernel_size", marker="o")

# Ensure x-axis only shows integer ticks
x_ticks = sorted(df_filtered[x].unique())
plt.xticks(x_ticks)
plt.yscale("log")
plt.xlabel(x)
plt.ylabel("Validation Loss")
plt.title("Validation Loss vs Residual Blocks (Grouped by Modes)")
plt.grid(False)
plt.tight_layout()
plt.show()
```
